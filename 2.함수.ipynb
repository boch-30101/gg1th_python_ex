### 함수 문제1 ###

```python
def mul(*values):
	mul = 1
	for i in values:
		mul *= i
	return mul
```

```python
print(mul(5,7,9,10))
print(mul(5,7,9,100))
```

```python
def comMul(num1, num2):
	# 1~100까지 공배수 구하기
	# 공배수 : 2 * 3의 배수
	# 1~100
	if (num1 % num2 == 0) or (num2 % num1 == 0):
		num = max(num1, num2)
	else:
		num = num1 * num2
	results = [] # 공배수 추가
	result = 0
	n = 1
	while result <= 100:
		result = num * n
		if result > 100:
			break
		results.append(result)
		n += 1
	return results
		
```

```python
input_data = input("두 수 입력 예) 2, 3 >")
input_data
```

```python
num1, num2 = input_data.split(",")
num1, num2 = int(num1), int(num2)
print(num1, num2)
```

```python
print(comMul(num1, num2))
```
