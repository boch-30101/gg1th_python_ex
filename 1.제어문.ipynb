```python
nums = [273, 310, 4, 6, 200, 100, 20, 50, 40, 101]
len(nums)
```

```python
# 100 이상의 숫자만 출력
for num in nums:
	# 100 이상 조건만 출력
	if num >= 100:
		print(f"-- 100이상의 수: {num}")
```

```python
#100이상의 숫자, 100미만인 숫자를 따로 새로운 리스트에 넣어서 각각 리스트로 출력
list_low = []
list_high = []
for num in nums:
	# 100 이상 조건만 출력
	if num >= 100:
		list_high.append(num)
	else:
		list_low.append(num)
print(f"- 100이상: {list_high}")
print(f"- 100미만: {list_low}")
```

```python
nums = [1,2,6,2,6,7,7,7]

counter = {}

for num in nums:
	# 키값 판단
	# if num in counter
	if counter.get(num) == None:
		# 그 숫자가 제일 처음 나왔을 때 -> 키값이 되어야함 + count +1
		counter[num] = 1
	# 이 숫자가 처음이 아닐 때 -> count +1
	else:
		counter[num] += 1

print(counter)
```

### 리스트 컴프리헨션 ###

```python
array = [i * i for i in range(0,20,2)]
array

# 예측결과 : 
```

```python
# array = [i * i for i in range(0,20,2)]
# for문으로 만들기

array = []

for i in range(0,20,2):
	array.append(i * i)
array 
```


### 리스트 컴프리헨션의 조건문 ###

```python
# 짝수만 출력하기
even_nums = []
for i in range(1,20+1):
	# i가 짝수인지 판단, 짝수이면 리스트에 추가
	if i % 2 == 0:
		even_nums.append(i)
even_nums
```

```python
array = [i for i in range(1,20+1) if i % 2 == 0]
array
```

### 제어문 종합 문제 ###

```python
decimal = []
for num in range(2,101):
	deno = num - 1
	while deno > 1:
		if num % deno == 0:
			break
		deno -= 1
	if deno == 1:
		decimal.append(num)
print(decimal)
```
